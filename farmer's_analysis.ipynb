## importing the excel file

```python
import pandas as pd
df=pd.read_excel("jpil_farmers_dataset.xlsx",header=1)
# print(df.head(1))
```

## total fertilizers used by 50 farmers

```python
farmers_fertilizers_analysis = df["Fertilizer Used (kg)"].sum()
print(f"the total fertilizers used by all farmers was {farmers_fertilizers_analysis}")
```

## fertilizers used by each village

```python
fertilizers_village = df.groupby("Village")["Fertilizer Used (kg)"].sum()
print(f"the amount of fertilizers used by each village was {fertilizers_village}")
```

## total farm size of all farmers

```python
total_farm_size = df["Farm Size (ha)"].sum()
print(f"the total farm size of all farmers in hectares was {total_farm_size}")
```

## Average Farm Size across all farmers

```python
Avg_farm_size = df["Farm Size (ha)"].mean().round()
print(f"the average farm size across all farmers is {Avg_farm_size}")
```

## Average yield after training in each village

```python
Avg_Yield_After = df.groupby("Village")["Yield After (kg/ha)"].mean().round(0)
print(f"the average yield after training in each {Avg_Yield_After}")
```

```python
import matplotlib.pyplot as plt
```

```python
#Creating a chart
Avg_Yield_After.plot(
    kind="bar",
    color=["red","green","red"],
    edgecolor="black"
)
plt.title("visualization of yield after training", fontsize=14)
plt.xlabel("village", fontsize=12)
plt.ylabel("Average Yield", fontsize=12)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Average amount of fertilizer used in each village

```python
Avg_Amount_Fertilizer = df.groupby("Village")["Fertilizer Used (kg)"].mean().round(0)
print(f"the amount of fertilizer used in each {Avg_Amount_Fertilizer}")
```

## The median (typical) yield change across all farmers 

```python
Median_Yield_Change = df["Yield Change (%)"].median()
print(f"the middle yield changes across all farmers {Median_Yield_Change}")
```

## the median output of farmers who used fertilizers and applied it correctly versus incorrectly

```python
Med_Fertilizer_Used = df.groupby("Fertilizer Type")["Fertilizer Used (kg)"].median()
print(f"the median fertilizers used by farmers correctly vs incorrectly {Med_Fertilizer_Used}")
```

## The Median Farm Size for cassava farmers compared to yam farmers

```python
Med_Farm_Size = df.groupby("Crop Type")["Farm Size (ha)"].median()
print(f"the median farm size for cassava vs yam farmers was {Med_Farm_Size}")
```

## The most common crop grown among 50 farmers

```python
Mode_Crop_Type = df["Crop Type"].mode().value_counts()
print(f"the most common crop type grown among all 50 farmers was {Mode_Crop_Type}")
```

## The village with the most farmers in the dataset

```python
Village_most_farmers = df["Village"].mode().value_counts()
print(f"the village with the most farmers in the dataset was {Village_most_farmers}")
```

## The most common land tenure type

```python
Land_Tenure = df["Land Tenure"].mode().value_counts()
print(f"the most common land tenure type is {Land_Tenure}")
```

## Checking if pest incidence is common or not among the farmers

```python
Pest_Incidence = df["Pest Incidence"].mode().value_counts()
print(f"check if pest incidence is more common among farmers {Pest_Incidence}")
```

## checking how most farmers applied fertilizers

```python
Fertilizer_Type = df["Fertilizer Type"].mode().value_counts()
print(f"did most of the farmers applied fertilizers correctly or incorrectly {Fertilizer_Type}")
```

## Checking to see if most farmers attended the JPIL training programme

```python
Attended_Training = df["Training Attended"].mode().value_counts()
print(f"check if most farmers attended the JPIL training programme {Attended_Training}")
```

## % of training attended

```python
Attended_Training = df["Training Attended"].value_counts()
Attended_Training.plot(
    kind="pie",
    color=["yellow", "red"],    
    autopct="%1.1f%%",
)
plt.title("percentage of farmers's training")
plt.tight_layout()
plt.show()
```

## How does the "Average Yield change(%) shift as farm size increases from small to large

```python
Average_Yield_Change = df.groupby("Farm Size (ha)")["Yield Change (%)"].mean().round(2)

Average_Yield_Change.plot(
    kind = "line",
    color = "blue",
    marker = "o",
    linewidth = 2
)
plt.title("a line graph showing yield change by farm size", fontsize=12)
plt.xlabel("Farm Size (ha)", fontsize=12)
plt.ylabel("Yield Chage (%)", fontsize=12)
plt.tight_layout()
plt.show()
```

## Scattered plot chart

```python
plt.figure(figsize=(10,7))
plt.scatter(
    df["Farm Size (ha)"],
    df["Fertilizer Used (kg)"],
    s = 15,
    alpha = 0.5
)
plt.title("village vs fertilizer's used", fontsize=14)
plt.xlabel("Farm Size", fontsize=12)
plt.ylabel("Fertilizer Used", fontsize=12)
plt.tight_layout()
plt.show()
```
